```python
from pathlib import Path

import matplotlib.image as mpimg
import matplotlib.pyplot as plt

def visualize_side_by_side(base):
    color_dir = base / "color"
    lines_dir = base / "lines"

    names = sorted(p.stem for p in color_dir.glob("*.png"))

    _, axes = plt.subplots(2, 5, figsize=(22, 11))

    for col, name in enumerate(names):
        # screenshots from open3d mesh
        ax = axes[0, col]
        ax.imshow(mpimg.imread(color_dir / f"{name}.png"))
        ax.set_xticks([])
        ax.set_yticks([])
        ax.set_xlabel(name, fontsize=13)

        # screenshots from pano annotator tool
        ax = axes[1, col]
        ax.imshow(mpimg.imread(lines_dir / f"{name}.png"))
        ax.set_xticks([])
        ax.set_yticks([])

    axes[0, 0].set_ylabel("color", fontsize=14)
    axes[1, 0].set_ylabel("lines", fontsize=14)

    plt.tight_layout()
    plt.show()
```
